## Goal
Build a chatbot using only built-in Python: dictionaries, strings, functions, conditionals, and loops. No packages or API key required.

```python
KNOWLEDGE = {
    "premium": "A premium is the amount you pay to keep an insurance policy active.",
    "deductible": "A deductible is an amount you may pay before covered benefits apply, depending on the policy.",
    "coverage": "Coverage describes what a policy includes, subject to its terms and exclusions.",
    "claim": "A claim is a request to an insurer for a benefit under a policy.",
}
ALIASES = {"premiums": "premium", "deductibles": "deductible", "claims": "claim"}

def chatbot_reply(message):
    normalized = message.strip().lower()
    if normalized in {"hello", "hi", "hey"}:
        return "Hello! Ask about premiums, deductibles, coverage, or claims."
    if normalized == "help":
        return "Topics: " + ", ".join(KNOWLEDGE)
    # Split into words so 'claim' does not accidentally match 'disclaimer'.
    words = normalized.translate(str.maketrans({c: " " for c in "?!.,:;"})).split()
    topics = {ALIASES.get(word, word) for word in words} & KNOWLEDGE.keys()
    if len(topics) > 1:
        return "Please ask about one topic at a time: " + ", ".join(sorted(topics))
    if topics:
        return KNOWLEDGE[next(iter(topics))]
    return "I don't have an answer for that. Type help to see my topics."
```

## Try predictable examples
Exact words and aliases work. Different wording such as “What do I pay each month?” needs another rule.

```python
for question in ["Hi", "What is a PREMIUM?", "Tell me about claims", "premium and deductible", "What do I pay each month?"]:
    print("You:", question)
    print("Bot:", chatbot_reply(question), "\n")

assert chatbot_reply(" PREMIUM ") == KNOWLEDGE["premium"]
assert chatbot_reply("claims?") == KNOWLEDGE["claim"]
assert "don't have" in chatbot_reply("disclaimer")
assert "one topic" in chatbot_reply("premium deductible")
print("All four behavior checks passed.")
```

## Chat interactively
Run this cell when you want to chat. Type `exit` to finish; use the stop button if necessary.

```python
def chat():
    print("Insurance bot: type help for topics or exit to stop.")
    while True:
        message = input("You: ")
        if message.strip().lower() == "exit":
            print("Bot: Goodbye!")
            break
        print("Bot:", chatbot_reply(message))

# Uncomment to start an interactive session:
# chat()
```
